## pandas (HTML output)

````python
import pandas as pd
df = pd.DataFrame({'city': ['Paris', 'Lyon', 'Nice'], 'population_k': [2102, 522, 342]})
df['share'] = (df.population_k / df.population_k.sum()).round(3)
df
````

## matplotlib (PNG output)

````python
import matplotlib.pyplot as plt
x = np.linspace(0, 2 * np.pi, 200)
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, np.sin(x), label='sin')
ax.plot(x, np.cos(x), label='cos')
ax.legend()
plt.close(fig)
fig
````
